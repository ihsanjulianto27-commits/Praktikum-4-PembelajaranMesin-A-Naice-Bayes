```python
# Import Libarary

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
data = pd.read_csv("drug200.csv")
data.head()
```

```python
data.info()
```

```python
data.describe()
```

```python
from numpy._core import numeric
# Mengambil kolom numerikal
numerical_features = [feature for feature in data.columns if data[feature].dtypes != 'O']
print("Jumlah Fitur Numerik = ",
      len (numerical_features))
print("Daftar Fitur Numerik",
      numerical_features)
display(data[numerical_features].head())
```

```python
from pandas.core.arrays import categorical
# mengambil kolom kategorikal
categorical_features = [feature for feature in data.columns if data[feature].dtypes == 'O']
print("Jumlah Fitur Kategorikal = ",
      len (categorical_features))
print("Daftar Fitur Kategorikal",
      categorical_features)
display(data[categorical_features].head())
```

```python
# Menampilkan data unik pada fitur kategorikal
for i in categorical_features:
  print(data[i].unique())
```

```python
# mengubah kolom kategorikal menjadi angka untuk klarifikasi
from sklearn.preprocessing import LabelEncoder
le = LabelEncoder()
for feature in categorical_features:
  data[feature] = le.fit_transform(data[feature])
print(data[feature].unique())
```

```python
x = data.drop(['Drug'], axis = 1)
y = data['Drug']
print(x)
print(y)
```

```python
from sklearn.model_selection import train_test_split
x_train, x_test, y_train, y_test = train_test_split(x, y, test_size = 0.2, random_state = 0)
```

```python
from sklearn.naive_bayes import GaussianNB
model = GaussianNB()
model.fit(x_train, y_train)
```

```python
#evaluasi Testing
y_pred_train = model.predict(x_train)
y_pred_test = model.predict(x_test)
```

```python
# Evaluasi Traning
from sklearn.metrics import confusion_matrix, classification_report

print("Evaluasi Traning : ")
cm_train = confusion_matrix(y_train, y_pred_train)
print(cm_train)
print(classification_report(y_train, y_pred_train))

plt.figure(figsize=(6,4))
sns.heatmap(cm_train, annot=True, fmt='d', cmap='Blues')
plt.title('Confusion Matrix Training')
plt.xlabel('Label Terprediksi')
plt.ylabel('Label Aktual')
plt.show()
```
